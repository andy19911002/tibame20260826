```python

def is_date_valid(d):
    day = d % 100
    remain = d // 100
    month = remain % 100
    year = remain // 100
    # print(year, month, day)

    if year >= 1:
        if 1 <= month <= 12:
            day_per_month = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
            day_max = day_per_month[month-1]
            if month == 2 and is_leap_year(year):
                day_max = 29
            if 1 <= day <= day_max:
                return True
            else:
                # day invalid
                return False
        else:
            # month invalid
            return False
    else:
        # year invalid
        return False

print(is_date_valid(20000229))
print(is_date_valid(20010229))
```

```python
# 想法1
# 日期往後增加, 字串->數字->開根號->是否整數
# 普通迴圈
import datetime as dt

def method_1(y=2024, m=10, d=1, max_count=10):
    d = dt.datetime(y, m, d)
    delta = dt.timedelta(days=1)
    count = 0
    result = []
    while True:
        d_int = int(d.strftime("%Y%m%d"))
        d_int_root = d_int ** 0.5
        if d_int_root.is_integer():
            result.append(d_int)
            count = count + 1
        if count == max_count:
            break
        d = d + delta
    return result

print(method_1())
print(method_1(max_count=5))
print(method_1(d=2))
```
